```python
# When we create a function inside the class then it is known as method and which are outside the class are known as just only Function.
# The main difference is where they(method and function) belong and how you call them.

# Function --> usually exists independently, called directly.
# Method --> belongs to an object or class, called through an object/class.
```

```python
name = "alice"

# Function
length = len(name)

# Method
uppercase = name.upper()

# simple way to remember
# Function = independent block of code.
# Method = function attached to an object or class, used to perform some specific operations on that object's data.
```

```python
name = "Reyansh"
print(name.upper())
# Here, upper() is a method of the string object name. It converts the string into uppercase.
```

```python
numbers = [1,2,3]
numbers.append(4)
numbers
# here, append() is a method of the list object. It adds 4 to the list.
```

```python
# why do we use methods ?
# Methods are useful because they allow an object to perform actions related to its own data

# car.start()
# car.stop()
# car.break()   these methods represent actions that 'car' object can perform.
```

```python
# Method in OOP --> methods are generally defined inside a class.

class Student:      # student -> class
    def study(self):      # study() -> method
        print("Student is studying.")

student1 = Student()    # student1 -> object
student1.study()        # calling the method
```

```python
# Identify which is function or method ?
L = [1,2,3,4,5]
len(L)         # Function because it is outside the list class.
L.append(5)     # append() is a function which is implemented inside the list class so it is a method.
```
