```python
# Now understand the concept of 'self'.
# as long as you create a methods so all of have a default parameter which is self.
# Golden rule of OOPs --> All the functions/methods which are inside the class is only access by the object of that class.
# The functions and data is inside the class is only and only access by object of that class.
# ID of object and ID os self both are point on the same memory address, so object is also known as self.
```

```python
# In OOPs, self refers to the current object of a class. It allows an object to access its own attributes and methods.
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def display(self):
        print(self.name)
        print(self.age)
s1 = Student("anuj", 20)
s1.display()
```

```python
# each object has its own data. self helps python identify which object's data is being accessed.
# self is used to access instance attributes and methods.
# self is a convention not a reserved keyword. you also use anything in place of self.
```
