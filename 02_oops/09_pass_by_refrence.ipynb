```python
class Person:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender

# Outside the class --> hence it is function
def greet(person):
    print('Hi my name is',person.name,'and I am a',person.gender)

p = Person('Anuj','male')
greet(p)     # here p is the object of Person class.

# That means you can also pass a object of your class as a input to the function.
```

```python
# In fact you do just opposite of this -> A function can give a object of class in return.

class Person:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender

# Outside the class --> hence it is function
def greet(person):
    print('Hi my name is',person.name,'and I am a',person.gender)
    p1 = Person('Ankit','male')
    return p1

p = Person('Anuj','male')
x = greet(p)
print(x.name)
print(x.gender)  # here a function return the object.
```

```python
# Now understand the mutability of object.
class Person:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender

# Outside the class --> hence it is function
def greet(person):
    print(id(person))
    print('Hi my name is',person.name,'and I am a',person.gender)
    p1 = Person('Ankit','male')
    return p1

p = Person('Anuj','male')
print(id(p))
greet(p)

# both have same memory address.
# So technically we understand when we pass the object to function as input so you don't send the object while you sent the reference/address of the object.
```

```python
class Person:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender

# Outside the class --> hence it is function
def greet(person):
    print(id(person))
    person.name = 'ankit'
    print(person.name)

p = Person('Anuj','male')
print(id(p))
greet(p)     # here p is the object of Person class.
print(p.name)

# understand what happen in the last line of code ?
# first p point to anuj and we sent p into the function 'person' and also person point the same value and using person you change the value from anuj to ankit.
# so p is also pointing to ankit.
```

```python
class Person:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender

# Outside the class --> hence it is function
def greet(person):
    person.name = 'ankit'
    return person

p = Person('Anuj','male')
print(id(p))
p1 = greet(p)     # here p is the object of Person class.
print(id(p1))

# Now object is mutable then address will be same other wise if immutable then address will change.
# so all the objects(objects of user define classes) in python are immutable.
```
