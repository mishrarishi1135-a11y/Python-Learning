```python
# Static variable and class variable both are same.

# In a large application, there would be lot of classes. for every entity, you built a class.
# Aggregation(Has-a-relationship) -> one class owns the other class, that means ones is owner and other is property.
# For example: Hotel owns a menu class.
```

```python
# Aggregation(Has-A-Relationship)
# Example
class Customer:
    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address   # There are so many things inside the address so we decide that creating owns address class.

    def print_address(self):
        print(self.address.city,self.address.pin,self.address.state)    # here Address class value is used by Customer class.

class Address:
    def __init__(self,city,pin,state):
        self.city = city
        self.pin = pin
        self.state = state

        # Now we create a object of customer class.
add1 = Address('Gurgaon',122011,'Haryana')
cst = Customer("Anuj",'male',add1)      # This scenario is known as aggregation.

cst.print_address()
```

```python
# what if we create address class variable private.
class Customer:
    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address

    def print_address(self):
        print(self.address.__city,self.address.pin,self.address.state)
class Address:
    def __init__(self,city,pin,state):
        self.__city = city
        self.pin = pin
        self.state = state
add1 = Address('Gurgaon',122011,'Haryana')
cst = Customer("Anuj",'male',add1)

# It shows a error because you can't access private attributes when you perform aggregation.
```

```python
# For this issue, we can use getter method
class Customer:
    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address

    def print_address(self):
        print(self.address.get_city(),self.address.pin,self.address.state)
class Address:
    def __init__(self,city,pin,state):
        self.__city = city
        self.pin = pin
        self.state = state

    def get_city(self):     # Now you can print the city value.
        return self.__city
add1 = Address('Gurgaon',122011,'Haryana')
cst = Customer("Anuj",'male',add1)

cst.print_address()
```

```python
# so own class does not access private variable of owned class.
# here Customer is owned class and address is own class.
```

```python
# How can we add method ?
class Customer:
    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address

    def print_address(self):
        print(self.address.get_city(),self.address.pin,self.address.state)

    def edit_profile(self,new_name,new_city,new_pin,new_state):
        self.name = new_name
        self.address.edit_address(new_city,new_pin,new_state)    # here we call direct edit_address function.
class Address:
    def __init__(self,city,pin,state):
        self.__city = city
        self.pin = pin
        self.state = state

    def get_city(self):
        return self.__city

    def edit_address(self,new_city,new_pin,new_state):
        self.__city = new_city
        self.pin = new_pin
        self.state = new_state
add1 = Address('Gurgaon',122011,'Haryana')
cst = Customer("Anuj",'male',add1)

cst.print_address()

cst.edit_profile("ankit","Bombay",123234,"Thane")
cst.print_address()

# name change hua customer class me but address change hua through edit_address inside the address class.
```
