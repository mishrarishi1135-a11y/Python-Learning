```python
# Now what is encapsulation ?
# encapsulation means building data and the methods that operate on data inside the class, while controlling to direct access to that data.

# In simple words, keep the data protected and provide controlled methods to access or modify it.
```

```python
# why we need encapsulation ?
# 1-> data hiding/data protection -> It prevents outside code from directly modifying important data.
# Like class decide itself how can modify the data.
```

```python
class Student:
    def __init__(self,marks):
        self.__marks = marks

    def set_marks(self,marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid Marks")
    def get_marks(self):
        return self.__marks
```

```python
s = Student(80)
s.set_marks(90)   # now we change the marks from 80 to 90.
print(s.get_marks())
```

```python
# But understand this and after this we can't set data more than 100. then it shows invalid marks.
s.set_marks(150)
```

```python
# Now understand the story of encapsulation:
class Atm:
    # constructor method is used to initialize the object of the class. using constructor we don't need to call the function.
    # as you create the object, constructor code automatically execute.
    def __init__(self):
        self.pin = ''
        self.balance = 0
        self.menu()

# add all functionality which are in the ATM
    def menu(self):
        user_input = input("""
        Hi how can I help you?
        1. press 1 to create pin
        2. press 2 to change pin
        3. press 3 to check balance
        4. press 4 to withdraw
        5. Anything else to exit
        """)
        if user_input == '1':
            self.create_pin()
            #create pin
        elif user_input == '2':
            self.change_pin()
            # change pin
        elif user_input == '3':
            self.check_balance()
            # check balance
        elif user_input == '4':
            self.withdraw()
            # withdraw
        else:
            exit()

    def create_pin(self):
        user_pin = input("enter your pin")
        self.pin = user_pin

        user_balance = int(input("enter balance"))
        self.balance = user_balance

        print("Pin created successfully.")
        self.menu()       # For continuity of the program we call self.menu() .

    def change_pin(self):
        old_pin = input("enter old pin")
        if old_pin == self.pin:
            # let him change the pin
            new_pin = input("enter new pin")
            self.pin = new_pin
            print("pin change successfully")
            self.menu()
        else:
            print("Don't allowed for this.")
            self.menu()

    def check_balance(self):
        user_pin = input("enter your pin")
        if user_pin == self.pin:
            print("your balance is ", self.balance)
        else:
            print("Go to hell")

    def withdraw(self):
        user_pin = input("enter the pin")
        if user_pin == self.pin:
            # allow to withdraw
            amount = int(input('enter the amount'))
            if amount <= self.balance:
                self.balance = self.balance - amount
                print("withdrawl successful and balance is ", self.balance)
            else:
                print("you did not allow for taking that amount.")  # If user withdraw more than balance.
        else:
            print("pin is not correct.")
        self.menu()
```

```python
# stroy -> Assume you have a account in SBI Bank and have some balance. you can deposit and withdraw of your balance.
# But anyone can directly change your balance data so for this problem we use the encapsulation.

# Code of complete BankAccount class
# Private attribute -> balance and account number.
# Public methods -> deposit, withdraw and get balance.
```

```python
# lets understand this problem without encapsulation.
class BankAccount:
    def __init__(self,balance):
        self.balance = balance
account = BankAccount(5000)
print(account.balance)
account.balance = -10000
print(account.balance)
```

```python
# So in our program, we have not any rule which prevent from negative balance.
# Python allows this operation because 'balance' is a normal public instance attribute.
# Now solve this problem using encapsulation.

# For understanding encapsulation: three main important points
# Public members, protected members and private members.
```

```python
# Public members:
class Student:
    def __init__(self,name):
        self.name = name
s1 = Student("Anuj")
print(s1.name)
```

```python
# here 'name' is public instance attributes and we can access it also can change.
s1.name = "Amit"
print(s1.name)
```

```python
# Protected Members(_variable)
# python me ek underscore se start hone wle names ko conventionally protected ya internal-use members mana jaa ta hai.
class Student:
    def __init__(self,name):
        self._name = name

s1 = Student("Rahul")
print(s1._name)
```

```python
# Understand this concept, In python like 'self.name' -> it is a public naming convention.
# But self._name -> internal-use ya self style naming convention hai.
```

```python
# private Members (start with double underscore)

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

account = BankAccount(50000)
print(account.__balance)

# this code show a attribute error.

# python internally store self.__balance as self._BankAccount__balance then in next cell
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

account = BankAccount(50000)
print(account._BankAccount__balance)

# so double underscore is also not complete security barrier.
```

```python
# Now we want a mechanism whose modify 'balance' through the only some valid operations.
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit Successful")
        else:
            print("Invalid amount")
    def withdraw(self,amount):
        if amount <= 0:
            print("Invalid amount")
        elif amount > self.__balance:
            print("Insufficient Balance")
        else:
            self.__balance -= amount
            print("Withdraw Successful")
    def get_balance(self):
        return self.__balance

account = BankAccount(5000)

account.deposit(2000)
account.withdraw(1000)

print(account.get_balance())
```

```python
# Setter -> use for updating the the values of attributes with controlled format.
class Student:
    def __init__(self,age):
        self.__age = age
    def get_age(self):
        return self.__age

    def set_age(self,age):
        if 0 < age < 120:
            self.__age = age
        else:
            print("Invalid age")

s1 = Student(20)
print(s1.get_age())

s1.set_age(21)
print(s1.get_age())

s1.set_age(-5)
```

```python
# static variables
# Using a instance(it's value is different for each object) variable you can't implement a counter.
# so here we need a static variable because it is a variable of class like instance is a variable of object.
# The value of static variable is same for each object.

# Like understand customer name, customer balance these are instance but what about Bank IFSC code ?
# Static because for each customer, same IFSc code
```

```python
# without a static variable
class Student:
    def __init__(self,name):
        self.name = name
        self.college = "SLIET"
s1 = Student("Rahul")
s2 = Student("Anuj")

print(s1.college)
print(s2.college)

# Here, college is an instance variable because self.college creates a separate attributes for every object
```

```python
# using a static(class) variable

class Student:
    college = "SLIET"   # class variable
    
    def __init__(self,name):
        self.name = name    # Instance variable

s1 = Student("Rahul")
s2 = Student("Anuj")

print(s1.college)
print(s2.college)
print(Student.college)
```

```python
# here college is a class variable shared by the objects.
# self.name = Instance variable(separate for each object)

# how can we modify a static variable ?
class Student:
    college = "Sliet"
s1 = Student()
s2 = Student()
Student.college = "IIT Delhi"     # static variable

print(s1.college)
print(s2.college)
```

```python
# If before name we write self then it is instance variable and before name we write class name then it is static variable.
```
