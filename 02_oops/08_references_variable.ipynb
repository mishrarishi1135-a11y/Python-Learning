```python
# reference variable

# reference variables hold the objects.
# we can create objects without reference variable as well
# An object can have multiple reference variables.
# Assigning a new reference variable to an existing object does not create a new object.
```

```python
# Object without a reference
class Person:
    def __init__(self):
        self.name = "anuj"
        self.gender = "male"

p = Person()
# p contains the address of object while p is not the object, p is a reference variable.
# TEchnically -> when we call Person() then create the object and we store it's reference in p.
```

```python
q = p # now p and q both are reference variable and pointing to the same memory location.
```

```python
print(id(p))
print(id(q))
```

```python
# Change attribute value with the help of 2nd object.
print(p.name)
print(q.name)
```

```python
q.name = "Ankit"
print(q.name)
print(p.name)
# You will have to be careful while creating the reference variable.
# Like q = p when you write that means you did not create a new object while you do change in existing object.
```
