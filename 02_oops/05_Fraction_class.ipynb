```python
class Fraction:

    # Parameterized constructor --> that constructor which need inputs(here x and y are inputs).
    def __init__(self,x,y):
        self.num = x
        self.den = y

    # 2nd magic method __str__ :
    def __str__(self):
        return '{}/{}'.format(self.num,self.den)
```

```python
fr1 = Fraction(3,4)
print(fr1)
```

```python
fr2 = Fraction(1,2)
print(fr2)
```

```python
print(fr1 + fr2)
```

```python
class Fraction:

    # Parameterized constructor --> that constructor which need inputs(here x and y are inputs).
    def __init__(self,x,y):
        self.num = x
        self.den = y

    # 2nd magic method __str__ :
    def __str__(self):
        return '{}/{}'.format(self.num,self.den)

    # 3rd magic method __add__ : automatic trigger as you put + operator in between two objects of that class.
    def __add__(self,other):    # when you add two fraction object then first is known as 'self' and other is known as 'other'
        new_num = self.num*other.den + other.num*self.den
        new_den = self.den*other.den

        return '{}/{}'.format(new_num,new_den)

    def __sub__(self,other):
        new_num = self.num*other.den - other.num*self.den
        new_den = self.den*other.den

        return '{}/{}'.format(new_num,new_den)

    def __mul__(self,other):
        new_num = self.num * other.num
        new_den = self.den * other.den

        return '{}/{}'.format(new_num,new_den)

    def __truediv__(self,other):    # for division
            new_num = self.num * other.den
            new_den = self.den * other.num
    
            return '{}/{}'.format(new_num,new_den)
    
```
