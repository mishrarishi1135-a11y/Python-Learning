```python
# How can object access attributes ?
# An object access it's attributes using the dot(.) operator.

class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

student1 = Student("Rahul",20)

# Here name and age are the attributes of the object.
```

```python
print(student1.name)
print(student1.age)
```

```python
# Important Distinction
# student1.name means access the attributes while student1.display() means access/call a method. Both use the dot(.) operator.
```

```python
class Person:
    def __init__(self,name_input,country_input):
        self.name = name_input
        self.country = country_input

    def greet(self):
        if self.country == "India":
            print("Namaste",self.name)
        else:
            print("Hello",self.name)
```

```python
# How to access attributes
p = Person("anuj","India")   # Here we create a object.
```

```python
p.country
```

```python
p.name
```

```python
# How to access methods -> for this we can attach a ().
p.greet()
```

```python
# What if I try to access non-existent attributes.
p.gender
# Person object has no attribute 'gender'.
```

```python
# Attribute creation inside the class from outside of the class (class ke bahar se attribute create kiya class ke andar).
p.gender = "male"
```

```python
p.gender
```
