```python
# Guessing Game application
import random
random.randint(1,100)
```

```python
jackpot = random.randint(1,100)
guess = int(input("Start guessing"))   # When we don't know the total no. of iteration then use while loop.

counter = 1
while guess != jackpot:
    if guess < jackpot:
        print("guess higher")
    else:
        print("guess lower")
    guess = int(input("Start guessing"))

    counter += 1
    print("correct answer")
    print("you took", counter,"attempts")
```

```python
# How to use nested loops (loop inside the loop is called nested loop)
# Time complexity --> O(n^2)  which is not good.

rows = int(input("enter the number of rows"))
for i in range(1, rows + 1):
    for j in range(0,i):
        print("*", end=" ")
    print("")
```

```python
# Break and continue statements.
# Break - will end the loop at that point
for i in range(1,11):
    if i == 5:
        break
    print(i)   # Now your loop is terminate at a specific condition.
```

```python
# Break statement is widely used in linear search for decrease the run time and get the output sooner.

# Continue statement: particular iteration skip ho Ja ta hai and code will run continue.
for i in range(1,11):
    if i == 5:
        continue
    print(i)

# use this continue statement in the e-commerce store for these products which are out of stock and we use continue statement.
```

```python
# Pass statement
for i in range(1,11):
    pass   # for waiting the future print statement.
```
