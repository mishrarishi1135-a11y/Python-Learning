```python
# Tuples --> it's basically very similar to lists.

# Learn how to create a tuple
T1 = ()
T1   # empty tuple
```

```python
# Homogenous tuple
T2 = (1,2,3,4,5)
T2
```

```python
# Heterogenous tuple --> all the items are different data types
T3 = ("Hello",4,5.6)
T3
```

```python
# You can also create a 2d or 3d tuple (Don't make a single item tuple because it will not be tuple it's a integer or other data type.)
t3 = (1,2,3,(4,5))
t3
```

```python
t4 = (5)
t4
```

```python
# see the it's type
type(t4)
```

```python
# How can we create a single item tuple, just see this
T5 = ("Hello")
type(T5)
```

```python
T5 = ("Hello",)
type(T5)
```

```python
T6 = tuple("Goa")
T6
```

```python
# Tuple from  a list
t6 = tuple([1,2,3,4,5])
t6
```
