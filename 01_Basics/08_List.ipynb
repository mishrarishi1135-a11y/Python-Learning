```python
# Title() --> it create capital of first letter all the words in a given string.
"hello how are you".title()
```

```python
# assume you have not this title() function so you write a customize code for this function which can perform the same work.
sample = "hello how are you"
# first access all the words using the split()
print(sample.split())
L = []
# run a loop for each element
for i in sample.split():
    print(i.capitalize())
    L.append(i.capitalize())
print(L)
print(" ".join(L))
```

```python
# Check for other output
sample = "today is raining day"
# first access all the words using the split()
print(sample.split())
L = []
# run a loop for each element
for i in sample.split():
    print(i.capitalize())
    L.append(i.capitalize())
print(L)
print(" ".join(L))
```
