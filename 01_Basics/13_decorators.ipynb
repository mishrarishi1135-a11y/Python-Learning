```python
# Decorators in python

# Adding things to something to make more attractive/presentable and this additional functionality is known as decorator.
# function --> we add some functionality in this --> and give it to decorator as a input --> return a modify function = Decorator(higher order function)
# Decorator --> it is a callable python object which can take function as a input which modifies other functions/class.
```

```python
def printer():
    print("welcome!")
    print("welcome!")

printer()
```

```python
# But i want print three times, can't modify this function(we have not access for changing in function)
# so we create a decorator function which adds functionality in this printer() function.
def decor(printer):
    def inner():
        printer()   #  call existing functionality
        # add new functionality
        print("welcome")
    return inner

def printer():
    print("welcome")
    print("welcome")

printer = decor(printer)
printer()

# so that we can modify the existing function.
```

```python
# You can also use this.
def decor(printer):
    def inner():
        printer()   #  call existing functionality
        # add new functionality
        print("welcome")
    return inner

@decor
def printer():
    print("welcome")
    print("welcome")

printer()
```

```python
# But we have another requirement in this function like it add three number so we can use decorator.
def decor(addition):
    def inner():
        result = addition()   # existing functionality.
        # add functionality here
        num3 = float(input("enter third number: "))
        result = result + num3
        return result
    return inner


    def addition():
        num1 = float(input("enter first number: "))
        num2 = float(input("enter second number: "))
    result = num1 + num2
    return result

addition = decor(addition)
print("addition is: ", addition())
```
