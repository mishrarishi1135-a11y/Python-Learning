```python
# Literals in python - this a raw data given in a variable (numeric, string, boolean and special literals).
a = 0b1010 # Binary literals
b = 100 # decimal literal
c = 0o310 # octal literal
d = 0x12c # Hexadecimal literal

# Float literal
float_1 = 1.5
# complex literal
x = 3.14j

print(a,b,c,d)
print(float_1)
print(x)
```

```python
# string literals
string = "This is a python"
strings = 'this is a python'
char = 'c'
unicode = u"\u001f600\u001F600\u0001F923"

print(string)
print(strings)
print(char)
print(unicode)

# For multiline string use this ("""" """").
```

```python
# Boolean literals
a = True + 4
b = False + 10
print("a:",a)
print("b:",b)  # here implicit type conversion.
```
