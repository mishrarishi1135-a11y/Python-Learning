```python
# here are some rules about dictionary:
# dictionary has no indexing.
# dictionary is a mutable data type.
# keys -> immutable, value -> they can be mutable.
# keys should be unique, do not repeat any key.

# Mutable data types --> list/ sets/ dictionary
# Immutable data types --> string/ tuples/ float/int /Boolean/ complex
```

```python
# create a empty dictionary
d = {}
d
```

```python
D = {"name":"Anuj","age":20}
D
```

```python
# Keys are not mutable, see this
d1 = {[1,2,3]:"Anuj"}
# unhashable type "list"
```

```python
d1 = {(1,2,3):"anuj"}   # Now use the tuple(immutable data types so it can be used as key).
d1
```

```python
# When you repeat the key more than one then it use only most recent(which will be write in last) key not other one.
d2 = {"name":"anuj","name":"rahul"}
d2
```

```python
# Dictionary inside the dictionary --> 2D dictionary
D3 = {"name":"rohit","college":"HIT","Marks":{"M1":56,"M2":89}}
D3
```

```python
# accessing item from the dictionary
# NOt use the index number, here we access by giving the key.
D3["college"]
```
