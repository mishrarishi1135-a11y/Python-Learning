```python
# sets -- it is a collection of unique elements. sets are unordered and automatically remove duplicates.
numbers = {1,2,3,4,5}
print(numbers)
```

```python
# Duplicates are removed automatically
numbers = {1,2,3,3,4,4,}
print(numbers)
```

```python
# Use a set when you want unique values and fast membership checks.
# sets have no indexing or slicing.
# seta don't allow mutable data types.
# sets itself a mutable data types.( so 2D and 3D are not possible)
```

```python
# create a set
s1 = {}   # It's not set.
s1
```

```python
type(s1)  # it's create a dictionary not set.
```

```python
# for creating a empty set
s2 = set()
s2
```

```python
type(s2)
```

```python
# Homogenous set
s3 = {1,2,3,4,5}
print(s3)
```

```python
# Heterogenous set
s4 = {1,2,3,"hello",5.6}
print(s4)
```

```python
# Sets do not allow the duplicates.
s3 = {1,1,3,4,5,5,5}
print(s3)
```

```python
# sets don't contain mutable data types.
s4 = {[1,2,3],"hello"}
print(s4)    # we don't write list inside a set because list are mutable data types
```

```python
# But we add tuple because tuples are immutable data types.
s4 = {(1,2,3),"hello"}
print(s4)
```

```python
# sets have no indexing, it follows the hashing technique by using this you decide that a particular data structure will go at which position.
# 2D sets are not possible
s5 = {{1,2},{3,4}}
print(s5)    # show error; unhashable type: 'set' both are mutable data types.
```

```python
s1 = {1,2,3,4,5}
print(s1[3])
# it show error, set object does not support indexing.( you can't access item from a set and no editing in items)
```
