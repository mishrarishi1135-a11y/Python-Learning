```python
# variable - the value which you don't know and it can fill in the future.
# How can create variable ?
# here we don't need to declare the variable and no need to tell data type which is known as Dynamic typing.
name = "Anuj"
print(name)
```

```python
# you can easily change the variable assign value. This feature is known as dynamic binding(store multiple data types)
name = 4
print(name)
```

```python
name = True
print(name)
```
