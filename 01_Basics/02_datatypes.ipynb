```python
# Python supports 3 categories of data types:
# Basic data types - (int, float, complex, boolean and string)
# Container data types - (List, Tuple, sets and dictionary)
# User-defined types - class
```

```python
# Integer
print(400000000000000000000000000000000000)  # How many large numbers it can print
```

```python
print(1e309)   # In this, it tells infinity and this is range of integer data type.
```

```python
# Float
print(4.5)
print(1.7e308)
```

```python
# But this is the range
print(1.7e309)
```

```python
# Boolean
print(True)
print(False)
```

```python
print(4+5j)  # also it allows complex numbers
```

```python
# String
print("Kolkata")
print('Kanpur')
print("""Delhi""")
```

```python
# List
print([1,2,3,4,5])
```

```python
# Tuple
print(1,2,3,4,5)
```

```python
# also we can write this form
print((1,2,3,4,5))
```

```python
# sets
print({1,2,3,4,5})
```

```python
# Dictionary -- Fast Searching
print({"name":"Anuj","age":"19","gender":"male"})
```
